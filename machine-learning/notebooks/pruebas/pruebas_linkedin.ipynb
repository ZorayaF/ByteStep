```python
# --- 1. Importación de Librerías Fundamentales ---
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("Librerías importadas exitosamente.")

# --- 2. Carga e Inspección de Datasets (Desde Raíz) ---

try:
    # --- A. Stack Overflow 2023 Survey ---
    # Asumiendo que el nombre del archivo es este
    df_so = pd.read_csv('datasets/stack overflow/stack_overflow_2023.csv')
    print(f"Stack Overflow cargado: {df_so.shape[0]} filas.")

    # --- B. O*NET Database (Archivos Clave) ---
    df_onet_occupation_data = pd.read_excel('datasets/ONET/Occupation Data.xlsx')
    df_onet_alternate_titles = pd.read_excel('datasets/ONET/Alternate Titles.xlsx')
    df_onet_tech_skills = pd.read_excel('datasets/ONET/Technology Skills.xlsx')
    df_onet_skills = pd.read_excel('datasets/ONET/Skills.xlsx')
    print("Archivos de O*NET cargados.")

    # --- C. LinkedIn Dataset (Archivos Clave con Corrección) ---

    # Archivo Central (con la corrección de errores)
    df_li_postings = pd.read_csv(
        'datasets/LinkedIn job posting/postings.csv',
        engine='python',
        on_bad_lines='skip'
    )
    print(f"LinkedIn Postings cargado: {df_li_postings.shape[0]} filas (filas con error omitidas).")

    # Archivos de LinkedIn (asumiendo que están todos en raíz)
    df_li_companies = pd.read_csv('datasets/LinkedIn job posting/companies.csv', on_bad_lines='skip')
    df_li_emp_counts = pd.read_csv('datasets/LinkedIn job posting/employee_counts.csv', on_bad_lines='skip')
    print("Archivos de LinkedIn 'datasets/LinkedIn job posting/companies' y 'employee_counts' cargados.")

    df_li_job_skills = pd.read_csv('datasets/LinkedIn job posting/job_skills.csv', on_bad_lines='skip')
    df_li_salaries = pd.read_csv('datasets/LinkedIn job posting/salaries.csv', on_bad_lines='skip')
    print("Archivos de LinkedIn 'job_skills' y 'salaries' cargados.")

    df_li_map_skills = pd.read_csv('datasets/LinkedIn job posting/skills.csv', on_bad_lines='skip')
    print("Archivo de LinkedIn 'skills' (mapeo) cargado.")

    print("\n¡Todos los archivos han sido cargados exitosamente!")

    # --- Paso 2 (Inicial): Inspección de Estructura y Llaves ---

    print("\n--- 1. Inspección de LinkedIn: Postings (El archivo central) ---")
    print(df_li_postings.info())

    print("\n--- 2. Inspección de LinkedIn: Salaries (Nuestro Objetivo 'Y') ---")
    print(df_li_salaries.info())

    print("\n--- 3. Inspección de LinkedIn: Job Skills (Nuestro Objetivo 'X') ---")
    print(df_li_job_skills.info())

    print("\n--- 4. Inspección de LinkedIn: Mapeo de Skills (El 'diccionario') ---")
    print(df_li_map_skills.info())

    print("\n--- 5. Inspección de LinkedIn: Companies (Contexto/Feature) ---")
    print(df_li_companies.info())

    print("\n--- 6. Inspección de LinkedIn: Employee Counts (Contexto/Feature) ---")
    print(df_li_emp_counts.info())

except FileNotFoundError as e:
    print("\n--- ¡ERROR DE RUTA! ---")
    print(f"No se encontró el archivo: {e.filename}. Asegúrate de que el nombre esté escrito exactamente igual (mayúsculas/minúsculas).")
except Exception as e:
    print(f"\nOcurrió un error inesperado durante la carga o inspección: {e}")
```

```python
try:
    print("--- 1. Explorando HABILIDADES (map_skills) ---")
    # 'map_skills' es pequeño (vimos 35 entradas), 
    # así que podemos imprimirlo todo.
    print(df_li_map_skills)

    print("\n--- 2. Explorando INDUSTRIAS (map_industries) ---")
    # 'map_industries' puede ser más grande. 
    # Imprimamos las primeras 100 para tener una idea.
    print(df_li_map_industries.head(100))
    
    # Si hay muchas, vamos a buscar palabras clave relevantes
    print("\n--- Búsqueda de industrias 'Tech' ---")
    
    # Re-creamos el diccionario de industrias si no existe
    if 'df_industries_full' not in locals():
        df_industries_full = pd.merge(df_li_job_industries, df_li_map_industries, on='industry_id', how='left')

    # Obtenemos TODOS los nombres de industrias únicos
    all_industries = df_industries_full['industry_name'].unique()
    
    print(f"Total de industrias únicas encontradas: {len(all_industries)}")
    
    # Buscamos las que nos interesan
    it_industry_keywords = [
        'Information', 'Software', 'Internet', 'Computer', 
        'Technology', 'Data', 'IT'
    ]
    industry_regex = '|'.join(it_industry_keywords)

    # Filtramos la lista de nombres únicos
    found_industries = [ind for ind in all_industries if pd.notna(ind) and pd.Series(ind).str.contains(industry_regex, case=False).any()]
    
    print("\nIndustrias candidatas para el filtro de TI:")
    print(found_industries)

except Exception as e:
    print(f"Ocurrió un error explorando: {e}")
```

```python
try:
    print("Iniciando el filtro preciso por HABILIDAD...")

    # --- A. Encontrar IDs por HABILIDAD de TI ---
    
    # 1. Definir nuestra lista de abreviaturas de TI (basada en tu salida)
    it_skill_abr = [
        'IT',    # Information Technology
        'ENG',   # Engineering
        'QA',    # Quality Assurance
        'ANLS',  # Analyst
        'SCI',   # Science
        'DSGN',  # Design
        'PRDM',  # Product Management
        'PRJM'   # Project Management
    ] 
    
    # 2. Crear el regex para buscar
    skill_regex = '|'.join(it_skill_abr)
    print(f"Buscando habilidades que coincidan con: {skill_regex}")

    # 3. Re-crear el diccionario de habilidades (job_id -> skill_abr)
    if 'df_skills_full' not in locals():
         df_skills_full = pd.merge(df_li_job_skills, df_li_map_skills, on='skill_abr', how='left')

    # 4. Filtrar los trabajos que tienen esas habilidades (usando skill_abr)
    it_skill_jobs = df_skills_full[
        df_skills_full['skill_abr'].str.contains(skill_regex, case=False, na=False)
    ]
    
    # 5. Obtener la lista única de job_id
    it_job_ids_from_skill = it_skill_jobs['job_id'].unique()
    print(f"Total de IDs de TI únicos (por habilidad): {len(it_job_ids_from_skill)}")

    
    # --- B. Aplicar el Filtro a 'Postings' ---
    
    # 1. Aplicar el filtro final a nuestra tabla principal de Postings
    df_master_it = df_li_postings[
        df_li_postings['job_id'].isin(it_job_ids_from_skill)
    ].copy()

    print("\n--- ¡FILTRO PRECISO (SOLO HABILIDAD) COMPLETADO! ---")
    print(f"Tamaño original de Postings: {df_li_postings.shape[0]}")
    print(f"Tamaño después del filtro de TI (df_master_it): {df_master_it.shape[0]}")

except Exception as e:
    print(f"Ocurrió un error durante el filtro preciso: {e}")
```

```python
# --- Paso 3 (Corregido): La Gran Fusión (versión IT) ---

try:
    # 1. Base DataFrame
    # Empezamos desde nuestro set filtrado (35,085 filas)
    df_master = df_master_it.copy() 
    print(f"1. Base inicial (postings IT): {df_master.shape}")

    # 2. Fusión de Salarios (Nuestro 'y')
    df_master = pd.merge(
        df_master,
        df_li_salaries,
        on='job_id',
        how='left',
        suffixes=('_post', '_sal') 
    )
    print(f"2. Después de Fusión de Salarios: {df_master.shape}") # Debe ser (35085, ...)

    # 3. Limpieza de Salarios
    sal_cols = ['max_salary', 'min_salary', 'med_salary', 'pay_period', 'currency', 'compensation_type']
    
    for col in sal_cols:
        df_master[col] = np.where(df_master[f'{col}_sal'].notnull(), df_master[f'{col}_sal'], df_master[f'{col}_post'])
        df_master = df_master.drop(columns=[f'{col}_post', f'{col}_sal'])
    
    print(f"3. Después de Limpieza de Salarios: {df_master.shape}") # Debe ser (35085, ...)

    # --- 4. Fusión de Información de Compañía (¡CORREGIDO!) ---
    
    # 4a. (LA CORRECCIÓN) De-duplicar la tabla de 'employee_counts'
    print("De-duplicando 'employee_counts' para obtener el más reciente...")
    
    # Ordenamos por fecha (el 'time_recorded' más alto es el más nuevo)
    df_emp_counts_sorted = df_li_emp_counts.sort_values('time_recorded', ascending=False)
    
    # Eliminamos duplicados en 'company_id', quedándonos con el 'first' (el más nuevo)
    df_emp_counts_unique = df_emp_counts_sorted.drop_duplicates(subset='company_id', keep='first')
    
    # Seleccionamos solo las columnas que queremos
    df_company_features = df_emp_counts_unique[['company_id', 'employee_count', 'follower_count']]

    # 4b. (LA FUSIÓN) Ahora unimos con la tabla 'unique'
    # Arreglamos los tipos de 'company_id' para que coincidan
    df_master['company_id'] = df_master['company_id'].astype('Int64')
    df_company_features['company_id'] = df_company_features['company_id'].astype('Int64')
    
    df_master = pd.merge(
        df_master,
        df_company_features,
        on='company_id',
        how='left'
    )
    print(f"4. Después de Fusión de Compañía: {df_master.shape}") # DEBE ser (35085, ...)

    # 5. Fusión de Habilidades (Features 'X')
    print("Agregando 'skills_list'...")
    if 'df_skills_full' not in locals():
        df_skills_full = pd.merge(df_li_job_skills, df_li_map_skills, on='skill_abr', how='left')
    
    df_skills_agg = df_skills_full.groupby('job_id')['skill_name'].apply(list).reset_index()
    df_skills_agg = df_skills_agg.rename(columns={'skill_name': 'skills_list'})
    
    df_master = pd.merge(
        df_master,
        df_skills_agg,
        on='job_id',
        how='left'
    )
    print(f"5. Después de Fusión de Habilidades: {df_master.shape}") # DEBE ser (35085, ...)
    
    # 6. Fusión de Industrias (Features 'X')
    print("Agregando 'industries_list'...")
    if 'df_industries_full' not in locals():
        df_industries_full = pd.merge(df_li_job_industries, df_li_map_industries, on='industry_id', how='left')
    
    df_industries_agg = df_industries_full.groupby('job_id')['industry_name'].apply(list).reset_index()
    df_industries_agg = df_industries_agg.rename(columns={'industry_name': 'industries_list'})

    df_master = pd.merge(
        df_master,
        df_industries_agg,
        on='job_id',
        how='left'
    )
    print(f"6. Después de Fusión de Industrias: {df_master.shape}") # DEBE ser (35085, ...)


    # --- Inspección Final ---
    print("\n¡GRAN FUSIÓN (CORREGIDA) COMPLETADA!")
    print(f"Dimensiones finales de 'df_master': {df_master.shape}")
    print("\nInspección del DataFrame 'df_master' (primeras filas y columnas clave):")
    
    key_cols = ['job_id', 'title', 'max_salary', 'pay_period', 'employee_count', 'skills_list', 'industries_list']
    print(df_master[key_cols].head())

except Exception as e:
    print(f"\nOcurrió un error durante la fusión: {e}")
```

```python
from flashtext import KeywordProcessor

try:
    print("Iniciando el escaneo granular con el diccionario O*NET...")

    # 1. Preparar el Diccionario (las 8,775 habilidades)
    # Obtenemos la lista única de habilidades, eliminamos NaNs,
    # las convertimos a string (por si acaso) y a minúsculas.
    skill_list = df_onet_tech_skills['Example'].dropna().astype(str).str.lower().unique().tolist()
    
    print(f"Total de habilidades únicas en el diccionario: {len(skill_list)}")

    # 2. Inicializar y "Armar" el procesador FlashText
    keyword_processor = KeywordProcessor()
    keyword_processor.add_keywords_from_list(skill_list)
    print("Procesador FlashText 'armado' con el diccionario O*NET.")

    # 3. Definir la función de escaneo
    def find_skills(description):
        # .extract_keywords() escanea el texto y devuelve una lista
        # de las palabras clave que encontró.
        # Convertimos la descripción a minúsculas para que coincida con el diccionario.
        return keyword_processor.extract_keywords(description.lower())

    # 4. Aplicar la función a las 35,085 descripciones
    print("Escaneando 35,085 descripciones... (Esto puede tardar 1-2 minutos)")
    
    # Rellenamos NaNs en 'description' por si acaso
    df_master['description'] = df_master['description'].fillna('')
    
    # .apply() ejecuta nuestra función 'find_skills' en cada fila
    df_master['granular_skills'] = df_master['description'].apply(find_skills)
    
    print("¡Escaneo completado! Nueva columna 'granular_skills' creada.")

    # 5. Inspeccionar los resultados
    print("\n--- Inspección de las Habilidades Encontradas ---")
    
    # Mostramos las filas que SÍ encontraron habilidades
    # (Filtramos las listas vacías '[]')
    print(df_master[df_master['granular_skills'].apply(lambda x: len(x) > 0)][['title', 'granular_skills']].head(10))

except Exception as e:
    print(f"Ocurrió un error: {e}")
    
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

try:
    print("Analizando la demanda de habilidades granulares...")

    # 1. Limpiar duplicados dentro de cada lista
    # Usamos set() para eliminar duplicados (ej. [php, php] -> [php])
    df_master['granular_skills_unique'] = df_master['granular_skills'].apply(
        lambda skills_list: list(set(skills_list))
    )

    # 2. "Explotar" el DataFrame
    # Esto toma cada habilidad en 'granular_skills_unique' y
    # crea una nueva fila para ella.
    df_exploded = df_master.explode('granular_skills_unique')

    # 3. Contar la frecuencia de cada habilidad
    # ¡Esta es la respuesta que hemos estado buscando!
    granular_demand = df_exploded['granular_skills_unique'].value_counts()

    print("\n--- Demanda de Habilidades Específicas (Top 20) ---")
    print(granular_demand.head(20))

    # --- 4. Visualización ---
    print("\nGenerando visualización de demanda granular (Top 20)...")
    
    # Tomamos las 20 habilidades principales para graficar
    top_20_skills = granular_demand.head(20)
    
    plt.figure(figsize=(12, 10))
    sns.barplot(
        x=top_20_skills.values, 
        y=top_20_skills.index,
        palette='viridis' # Una paleta de colores diferente
    )
    plt.title('Top 20 Habilidades Específicas Más Demandadas', fontsize=16)
    plt.xlabel('Número de Ofertas de Trabajo', fontsize=12)
    plt.ylabel('Habilidad (O*NET)', fontsize=12)
    plt.show()

except Exception as e:
    print(f"Ocurrió un error: {e}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

try:
    print("Creando DataFrame binario para el Top 20...")

    # 1. Obtener la lista de los 20 nombres de habilidades principales
    top_20_skill_names = granular_demand.head(20).index.tolist()

    # 2. Crear un DataFrame binario (One-Hot)
    # Usaremos una list comprehension para esto, es muy eficiente.
    
    # DataFrame temporal para nuestro mapa de calor
    df_heatmap = pd.DataFrame()

    # 'apply' itera sobre cada lista de habilidades en la columna
    for skill in top_20_skill_names:
        df_heatmap[skill] = df_master['granular_skills_unique'].apply(
            lambda skills_list: 1 if skill in skills_list else 0
        )
    
    print("DataFrame binario creado. Calculando correlación...")

    # 3. Calcular la matriz de correlación
    corr_matrix = df_heatmap.corr()

    # --- 4. Visualización ---
    print("Generando mapa de calor de correlación...")

    plt.figure(figsize=(15, 12)) # Hacemos la figura más grande
    sns.heatmap(
        corr_matrix,
        annot=True,           # Mostrar los números
        fmt='.2f',            # 2 decimales
        cmap='Blues',         # Paleta de colores
        xticklabels=top_20_skill_names,
        yticklabels=top_20_skill_names
    )
    
    plt.title('Correlación entre las Top 20 Habilidades Específicas', fontsize=18)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout() # Asegura que las etiquetas no se corten
    plt.show()

except Exception as e:
    print(f"Ocurrió un error: {e}")
    print("Asegúrate de que 'df_master' y 'granular_demand' existan.")
```
